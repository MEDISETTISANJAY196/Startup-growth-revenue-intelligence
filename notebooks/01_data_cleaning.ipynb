```python
customers = pd.read_csv("../data/raw/customers.csv")
products = pd.read_csv("../data/raw/products.csv")
orders = pd.read_csv("../data/raw/orders.csv")
marketing = pd.read_csv("../data/raw/marketing_campaigns.csv")
interactions = pd.read_csv("../data/raw/customer_interactions.csv")
expenses = pd.read_csv("../data/raw/expenses.csv")
returns = pd.read_csv("../data/raw/returns.csv")

print("All datasets loaded successfully!")
```

```python
import pandas as pd
import numpy as np
```

```python
# Store all datasets in a dictionary

datasets = {
    "Customers": customers,
    "Products": products,
    "Orders": orders,
    "Marketing": marketing,
    "Interactions": interactions,
    "Expenses": expenses,
    "Returns": returns
}

# Display basic information about each dataset

for name, df in datasets.items():
    print("=" * 60)
    print(f"DATASET: {name}")
    print("=" * 60)
    
    print("\nShape:")
    print(df.shape)
    
    print("\nColumns:")
    print(df.columns.tolist())
    
    print("\nData Types:")
    print(df.dtypes)
    
    print("\nMissing Values:")
    print(df.isnull().sum())
    
    print("\nDuplicate Rows:")
    print(df.duplicated().sum())
    
    print("\nFirst 5 Rows:")
    print(df.head())
    
    print("\n")
```

```python
# Create a copy of the customers dataset
customers_clean = customers.copy()

# Check duplicates before cleaning
print("Duplicates before cleaning:", customers_clean.duplicated().sum())

# Remove duplicate rows
customers_clean = customers_clean.drop_duplicates()

# Fill missing age values with median age
customers_clean["age"] = customers_clean["age"].fillna(
    customers_clean["age"].median()
)

# Convert signup_date to datetime
customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"]
)

# Check data after cleaning
print("\nDuplicates after cleaning:", customers_clean.duplicated().sum())
print("Missing values after cleaning:")
print(customers_clean.isnull().sum())

print("\nCleaned Customers Shape:", customers_clean.shape)
```

```python
# -----------------------------
# PRODUCTS CLEANING
# -----------------------------

products_clean = products.copy()

print("PRODUCTS")
print("Duplicates before:", products_clean.duplicated().sum())

products_clean = products_clean.drop_duplicates()

# Check missing values
print("Missing values:")
print(products_clean.isnull().sum())

print("Duplicates after:", products_clean.duplicated().sum())
print("Shape:", products_clean.shape)


# -----------------------------
# ORDERS CLEANING
# -----------------------------

orders_clean = orders.copy()

print("\n" + "=" * 50)
print("ORDERS")
print("Duplicates before:", orders_clean.duplicated().sum())

orders_clean = orders_clean.drop_duplicates()

# Fill missing payment methods
orders_clean["payment_method"] = orders_clean["payment_method"].fillna(
    "Unknown"
)

# Convert order_date to datetime
orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"]
)

print("Missing values after cleaning:")
print(orders_clean.isnull().sum())

print("Duplicates after:", orders_clean.duplicated().sum())
print("Shape:", orders_clean.shape)


# -----------------------------
# MARKETING CLEANING
# -----------------------------

marketing_clean = marketing.copy()

print("\n" + "=" * 50)
print("MARKETING")

marketing_clean = marketing_clean.drop_duplicates()

marketing_clean["campaign_date"] = pd.to_datetime(
    marketing_clean["campaign_date"]
)

print("Missing values:")
print(marketing_clean.isnull().sum())

print("Duplicates:", marketing_clean.duplicated().sum())
print("Shape:", marketing_clean.shape)


# -----------------------------
# CUSTOMER INTERACTIONS CLEANING
# -----------------------------

interactions_clean = interactions.copy()

print("\n" + "=" * 50)
print("CUSTOMER INTERACTIONS")

interactions_clean = interactions_clean.drop_duplicates()

interactions_clean["interaction_date"] = pd.to_datetime(
    interactions_clean["interaction_date"]
)

print("Missing values:")
print(interactions_clean.isnull().sum())

print("Duplicates:", interactions_clean.duplicated().sum())
print("Shape:", interactions_clean.shape)


# -----------------------------
# EXPENSES CLEANING
# -----------------------------

expenses_clean = expenses.copy()

print("\n" + "=" * 50)
print("EXPENSES")

expenses_clean = expenses_clean.drop_duplicates()

expenses_clean["expense_date"] = pd.to_datetime(
    expenses_clean["expense_date"]
)

print("Missing values:")
print(expenses_clean.isnull().sum())

print("Duplicates:", expenses_clean.duplicated().sum())
print("Shape:", expenses_clean.shape)


# -----------------------------
# RETURNS CLEANING
# -----------------------------

returns_clean = returns.copy()

print("\n" + "=" * 50)
print("RETURNS")

returns_clean = returns_clean.drop_duplicates()

returns_clean["return_date"] = pd.to_datetime(
    returns_clean["return_date"]
)

print("Missing values:")
print(returns_clean.isnull().sum())

print("Duplicates:", returns_clean.duplicated().sum())
print("Shape:", returns_clean.shape)
```

```python
# Save all cleaned datasets

customers_clean.to_csv(
    "../data/cleaned/customers_cleaned.csv",
    index=False
)

products_clean.to_csv(
    "../data/cleaned/products_cleaned.csv",
    index=False
)

orders_clean.to_csv(
    "../data/cleaned/orders_cleaned.csv",
    index=False
)

marketing_clean.to_csv(
    "../data/cleaned/marketing_campaigns_cleaned.csv",
    index=False
)

interactions_clean.to_csv(
    "../data/cleaned/customer_interactions_cleaned.csv",
    index=False
)

expenses_clean.to_csv(
    "../data/cleaned/expenses_cleaned.csv",
    index=False
)

returns_clean.to_csv(
    "../data/cleaned/returns_cleaned.csv",
    index=False
)

print("All cleaned datasets saved successfully!")
```
